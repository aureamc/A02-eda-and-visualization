$$

$$

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


# =============================================================================
# Q2.1: Load the dataset
# =============================================================================
def load_data(file_path):
    """Loads a CSV file into a pandas DataFrame."""
    try:
        df = pd.read_csv(file_path)
        print(f"Successfully loaded data from {file_path}. Shape: {df.shape}")
        return df
    except FileNotFoundError:
        print(f"Error: The file {file_path} was not found. Please ensure it's in the correct directory.")
        return None

df = load_data('./data/ForeignGifts_edu.csv')


if df is not None:
    # =============================================================================
    # Q2.2: Foreign Gift Amount histogram and description
    # =============================================================================
    print("\n--- Q2.2: Foreign Gift Amount histogram and description ---")
    # Convert 'Foreign Gift Amount' to numeric, handling potential non-numeric entries
    df['Foreign Gift Amount'] = pd.to_numeric(df['Foreign Gift Amount'], errors='coerce')
    df_q2 = df.dropna(subset=['Foreign Gift Amount']).copy() # Create a copy for analysis

    print("Descriptive statistics for 'Foreign Gift Amount':")
    print(df_q2['Foreign Gift Amount'].describe())

    plt.figure(figsize=(10, 6))
    sns.histplot(df_q2['Foreign Gift Amount'], bins=50, kde=True)
    plt.title('Distribution of Foreign Gift Amount')
    plt.xlabel('Foreign Gift Amount')
    plt.ylabel('Frequency')
    plt.yscale('log') # Use log scale for y-axis due to wide range of values
    plt.show()
    print("\nFindings for Foreign Gift Amount: The distribution is highly skewed to the right, with most gifts being small amounts and a few very large gifts. A log scale on the y-axis helps visualize the frequency of smaller amounts. The mean is significantly higher than the median, confirming the right skewness.")


    # =============================================================================
    # Q2.3: Gift Type histogram/value counts and proportion
    # =============================================================================
    print("\n--- Q2.3: Gift Type histogram/value counts and proportion ---")
    print("Value counts for 'Gift Type':")
    gift_type_counts = df_q2['Gift Type'].value_counts()
    print(gift_type_counts)

    plt.figure(figsize=(12, 7))
    gift_type_counts.plot(kind='bar')
    plt.title('Distribution of Gift Types')
    plt.xlabel('Gift Type')
    plt.ylabel('Count')
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.show()

    total_gifts = gift_type_counts.sum()
    contract_prop = gift_type_counts.get('Contract', 0) / total_gifts
    real_estate_prop = gift_type_counts.get('Real Estate', 0) / total_gifts
    monetary_prop = gift_type_counts.get('Monetary Gift', 0) / total_gifts

    print(f"Proportion of Contracts: {contract_prop:.2%}")
    print(f"Proportion of Real Estate: {real_estate_prop:.2%}")
    print(f"Proportion of Monetary Gifts: {monetary_prop:.2%}")

    print("\nFindings for Gift Type: Monetary Gifts are the most common type, followed by Contracts. Real Estate makes up a smaller proportion.")


    # =============================================================================
    # Q2.4: Kernel density plots of log of Foreign Gift Amount
    # =============================================================================
    print("\n--- Q2.4: Kernel density plots of log of Foreign Gift Amount ---")
    # Handle zero or negative gift amounts if any, by adding a small constant or filtering
    df_q2['Log Foreign Gift Amount'] = np.log1p(df_q2['Foreign Gift Amount'])

    plt.figure(figsize=(10, 6))
    sns.kdeplot(df_q2['Log Foreign Gift Amount'], fill=True)
    plt.title('Kernel Density Plot of Log(1 + Foreign Gift Amount)')
    plt.xlabel('Log(1 + Foreign Gift Amount)')
    plt.ylabel('Density')
    plt.show()

    plt.figure(figsize=(12, 7))
    sns.kdeplot(data=df_q2, x='Log Foreign Gift Amount', hue='Gift Type', fill=True, common_norm=False)
    plt.title('Kernel Density Plot of Log(1 + Foreign Gift Amount) by Gift Type')
    plt.xlabel('Log(1 + Foreign Gift Amount)')
    plt.ylabel('Density')
    plt.legend(title='Gift Type')
    plt.show()

    print("\nPatterns in Log Foreign Gift Amount: The log transformation makes the distribution more symmetrical, closer to a normal distribution. Conditional on gift type, it's evident that different gift types have distinct distributions of amounts. For instance, 'Monetary Gift' and 'Contract' might have peaks at different log amounts, suggesting typical gift sizes vary by type. 'Real Estate' might show a different pattern as well.")


    # =============================================================================
    # Q2.5: Top 15 countries by number of gifts and amount given
    # =============================================================================
    print("\n--- Q2.5: Top 15 countries by number of gifts and amount given ---")
    top_countries_by_count = df_q2['Foreign Country Name'].value_counts().head(15)
    print("\nTop 15 countries by number of gifts:")
    print(top_countries_by_count)

    top_countries_by_amount = df_q2.groupby('Foreign Country Name')['Foreign Gift Amount'].sum().nlargest(15)
    print("\nTop 15 countries by amount given:")
    print(top_countries_by_amount)

    print("\nFindings for Top Countries: The top countries by number of gifts may not necessarily be the same as the top countries by total amount given. Some countries might give many small gifts, while others give fewer but larger gifts.")


    # =============================================================================
    # Q2.6: Giftors providing the most money in total
    # =============================================================================
    print("\n--- Q2.6: Giftors providing the most money in total ---")
    top_giftors_by_amount = df_q2.groupby('Foreign Giftor Name')['Foreign Gift Amount'].sum().nlargest(10) # Showing top 10 as 15 might be too long
    print("\nTop Giftors providing the most money in total:")
    print(top_giftors_by_amount)

    print("\nFindings for Top Giftors: This shows the specific entities (individuals or organizations) that have contributed the largest total amounts.")
```
